````python
from scipy.stats import chi2_contingency
import pandas as pd

categorical_cols = [
    'Sex', 'SibSp', 'Parch',
    'Class_1', 'Class_2', 'Class_3',
    'Embark_0', 'Embark_1', 'Embark_2',
    'Age_category_Young Adult', 'Age_category_Adult', 'Age_category_Senior'
]

results = []

for col in categorical_cols:
    contingency = pd.crosstab(df_use[col], df_use['Survived'])
    chi2, p, dof, expected = chi2_contingency(contingency)
    results.append({
        'Feature': col,
        'Chi2': chi2,
        'p-value': p,
        'Significant (p<0.05)': p < 0.05
    })

chi_square_df = pd.DataFrame(results).sort_values('p-value')
chi_square_df
````

### Preprocessing `df2`

````python
# 1. Create family_size column
df2['family_size'] = df2['SibSp'] + df2['Parch'] + 1

# 2. Create Age_category and one-hot encode it
# Fill any remaining NaN in Age before categorization, using the median as per earlier steps
median_age_df2 = df2['Age'].median()
df2['Age'] = df2['Age'].fillna(median_age_df2)

df2['Age_category'] = pd.cut(
    df2['Age'],
    bins=[0, 15, 25, 65, 122],
    labels=['Child', 'Young Adult', 'Adult', 'Senior'],
    right=False
)
df2 = pd.get_dummies(df2, columns=['Age_category'], prefix='Age_category', drop_first=False)
````

````python
# 3. Create a column with name prefix (Mr, Mrs, Miss, Master, etc.) and one-hot encode it
import re
def get_title(name):
    title_search = re.search(' ([A-Za-z]+)\.', name)
    if title_search:
        return title_search.group(1)
    return ""

df2['Name_prefix'] = df2['Name'].apply(get_title)

# Group rare titles
df2['Name_prefix'] = df2['Name_prefix'].replace(['Lady', 'Countess','Capt', 'Col', 'Don', 'Dr', 'Major', 'Rev', 'Sir', 'Jonkheer', 'Dona'], 'Rare')
df2['Name_prefix'] = df2['Name_prefix'].replace('Mlle', 'Miss')
df2['Name_prefix'] = df2['Name_prefix'].replace('Ms', 'Miss')
df2['Name_prefix'] = df2['Name_prefix'].replace('Mme', 'Mrs')

df2 = pd.get_dummies(df2, columns=['Name_prefix'], prefix='Name_prefix', drop_first=False)
````

````python
# 4. One-hot encode Pclass and Embarked
# Before one-hot encoding, fill any missing 'Embarked' values
most_frequent_embarked_df2 = df2['Embarked'].mode()[0]
df2['Embarked'] = df2['Embarked'].fillna(most_frequent_embarked_df2)

df2 = pd.get_dummies(df2, columns=['Pclass'], prefix='Class', drop_first=False)
df2 = pd.get_dummies(df2, columns=['Embarked'], prefix='Embark', drop_first=False)
````

````python
# 5. Drop the Ticket column, Name column, and original SibSp/Parch if family_size is used instead
df2 = df2.drop(['Ticket', 'Name'], axis=1)
````

````python
# Convert boolean columns created by get_dummies to int
for col in df2.columns:
    if df2[col].dtype == 'bool':
        df2[col] = df2[col].astype(int)

# Display updated df2 info and head
df2.info()
display(df2.head())
````

````python
# 6. Standard scale numerical columns
from sklearn.preprocessing import StandardScaler

# Identify numerical columns to scale
cols_to_scale_df2 = ['Age', 'Fare', 'family_size']

# Initialize StandardScaler
scaler_df2 = StandardScaler()

# Apply scaling to the selected columns in df2
df2[cols_to_scale_df2] = scaler_df2.fit_transform(df2[cols_to_scale_df2])

display(df2.head())
````

````python
from scipy.stats import chi2_contingency
import pandas as pd

categorical_cols = [
    'Sex',
    'Age_category_Child', 'Age_category_Young Adult', 'Age_category_Adult', 'Age_category_Senior',
    'Name_prefix_Master', 'Name_prefix_Miss', 'Name_prefix_Mr', 'Name_prefix_Mrs', 'Name_prefix_Rare',
    'Class_1', 'Class_2', 'Class_3',
    'Embark_C', 'Embark_Q', 'Embark_S'
]

results = []

for col in categorical_cols:
    contingency = pd.crosstab(df2[col], df2['Survived'])
    chi2, p, dof, expected = chi2_contingency(contingency)
    results.append({
        'Feature': col,
        'Chi2': chi2,
        'p-value': p,
        'Significant (p<0.05)': p < 0.05
    })

chi_square_df = pd.DataFrame(results).sort_values('p-value')
chi_square_df
````

````python
import re
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

# ============================================================
# STEP 1: Load and preprocess TRAINING data
# ============================================================
train = pd.read_csv("/content/drive/MyDrive/Datasets/train.csv")

def get_title(name):
    match = re.search(r' ([A-Za-z]+)\.', name)
    return match.group(1) if match else ""

def preprocess(df, median_age, mode_embarked):
    df = df.copy()

    # Family size
    df['family_size'] = df['SibSp'] + df['Parch'] + 1

    # Age: fill with given median, then bucket
    df['Age'] = df['Age'].fillna(median_age)
    df['Age_category'] = pd.cut(
        df['Age'], bins=[0, 15, 25, 65, 122],
        labels=['Child', 'Young Adult', 'Adult', 'Senior'], right=False
    )
    df = pd.get_dummies(df, columns=['Age_category'], prefix='Age_category')

    # Title / Name prefix
    df['Name_prefix'] = df['Name'].apply(get_title)
    df['Name_prefix'] = df['Name_prefix'].replace(
        ['Lady', 'Countess', 'Capt', 'Col', 'Don', 'Dr', 'Major', 'Rev', 'Sir', 'Jonkheer', 'Dona'], 'Rare'
    )
    df['Name_prefix'] = df['Name_prefix'].replace({'Mlle': 'Miss', 'Ms': 'Miss', 'Mme': 'Mrs'})
    df = pd.get_dummies(df, columns=['Name_prefix'], prefix='Name_prefix')

    # Sex
    df['Sex'] = df['Sex'].map({'male': 0, 'female': 1})

    # Embarked and Pclass
    df['Embarked'] = df['Embarked'].fillna(mode_embarked)
    df = pd.get_dummies(df, columns=['Pclass'], prefix='Class')
    df = pd.get_dummies(df, columns=['Embarked'], prefix='Embark')

    # Fare: fill any missing (test.csv has one)
    if df['Fare'].isnull().sum() > 0:
        df['Fare'] = df['Fare'].fillna(df['Fare'].median())

    # Drop unused columns
    df = df.drop(['Ticket', 'Name'], axis=1)

    # Bool -> int
    for col in df.columns:
        if df[col].dtype == 'bool':
            df[col] = df[col].astype(int)

    return df

# Compute stats from TRAINING data only
median_age = train['Age'].median()
mode_embarked = train['Embarked'].mode()[0]

train_processed = preprocess(train, median_age, mode_embarked)

# Feature list for model x03
feature_cols = ['Sex', 'Fare', 'family_size', 'Age_category_Child',
                 'Name_prefix_Master', 'Name_prefix_Miss', 'Name_prefix_Mr', 'Name_prefix_Mrs',
                 'Class_1', 'Class_2', 'Class_3', 'Embark_C', 'Embark_S']

X = train_processed[feature_cols].copy()
y = train_processed['Survived']

# Split for validation (optional, just to confirm training worked)
X_train, X_val, y_train, y_val = train_test_split(X, y, test_size=0.20, random_state=42)

# Fit scaler on TRAINING split only
scaler = StandardScaler()
X_train[['Fare', 'family_size']] = scaler.fit_transform(X_train[['Fare', 'family_size']])
X_val[['Fare', 'family_size']] = scaler.transform(X_val[['Fare', 'family_size']])

# Train final model with x03's best hyperparameters
model = RandomForestClassifier(
    max_depth=5, max_features='sqrt', min_samples_leaf=2,
    min_samples_split=2, n_estimators=200, random_state=42
)
model.fit(X_train, y_train)
print("Validation accuracy:", accuracy_score(y_val, model.predict(X_val)))

# Refit on ALL training data before predicting on test (use every row you have)
X_full_scaled = X.copy()
X_full_scaled[['Fare', 'family_size']] = scaler.fit_transform(X[['Fare', 'family_size']])
model.fit(X_full_scaled, y)

# ============================================================
# STEP 2: Load and preprocess TEST data (using TRAIN's stats)
# ============================================================
test = pd.read_csv("/content/drive/MyDrive/Datasets/test.csv")
test_processed = preprocess(test, median_age, mode_embarked)  # reuse train's median/mode

# Ensure test has every column the model expects (fill missing dummy cols with 0)
for col in feature_cols:
    if col not in test_processed.columns:
        test_processed[col] = 0

X_test_final = test_processed[feature_cols].copy()
X_test_final[['Fare', 'family_size']] = scaler.transform(X_test_final[['Fare', 'family_size']])

# ============================================================
# STEP 3: Predict and build submission file
# ============================================================
predictions = model.predict(X_test_final)

submission = pd.DataFrame({
    'PassengerId': test['PassengerId'],
    'Survived': predictions
})
submission.to_csv('submission.csv', index=False)
submission.head()
````
